# Notebook with Injected Parameters

```
# Parameters cell
batch_date = '2026-01-01'
batch_size = 100
env = 'dev'
```

```
print(f'Running with batch_date={batch_date}, batch_size={batch_size}, env={env}')
import duckdb
con = duckdb.connect()
df = con.execute('SELECT ? as batch_date, ? as batch_size, ? as env', [batch_date, batch_size, env]).fetchdf()
print(df)
```
